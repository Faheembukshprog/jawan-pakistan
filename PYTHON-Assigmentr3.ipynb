```python
# Task 1: Calculator
val1 = float(input("Enter first number: "))
val2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /, **): ")

if operator == '+':
    print(f"Result: {val1 + val2}")
elif operator == '-':
    print(f"Result: {val1 - val2}")
elif operator == '*':
    print(f"Result: {val1 * val2}")
elif operator == '/':
    if val2 != 0:
        print(f"Result: {val1 / val2}")
    else:
        print("Error: Division by zero is not allowed.")
elif operator == '**':
    print(f"Result: {val1 ** val2}")
else:
    print("Invalid operator entered.")
```

```python
# Task 2: Check for numeric value in list
mixed_list = ["Apple", "Orange", 45, "Banana", 10.5, True]
found = False

for item in mixed_list:
    if isinstance(item, (int, float)) and not isinstance(item, bool):
        print(f"Found numeric value: {item}")
        found = True

if not found:
    print("No numeric values found in the list.")
```

```python
# Task 3: Add a key to a dictionary
customer_data = {"name": "Ahmed", "city": "Karachi"}
print(f"Original Dictionary: {customer_data}")

# Adding a new key 'email'
customer_data["email"] = "ahmed@example.com"
print(f"Updated Dictionary: {customer_data}")
```

```python
# Task 4: Sum all numeric items in a dictionary
inventory = {
    "apples": 50,
    "oranges": "out of stock",
    "bananas": 30,
    "grapes": 20,
    "status": "active"
}

total_sum = sum(value for value in inventory.values() if isinstance(value, (int, float)))
print(f"The sum of all numeric items is: {total_sum}")
```

```python
# Task 5: Identify duplicate values from list
numbers = [1, 2, 3, 4, 2, 7, 8, 8, 3]
duplicates = []
unique_items = []

for num in numbers:
    if num not in unique_items:
        unique_items.append(num)
    else:
        if num not in duplicates:
            duplicates.append(num)

print(f"Duplicate values found: {duplicates}")
```

```python
# Task 6: Check if a given key already exists
student = {"id": 101, "name": "Sara", "course": "Python"}

key_to_check = "name"

if key_to_check in student:
    print(f"Yes, the key '{key_to_check}' exists in the dictionary.")
else:
    print(f"No, the key '{key_to_check}' does not exist.")
```
